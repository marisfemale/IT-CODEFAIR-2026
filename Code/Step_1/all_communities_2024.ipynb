```python
import geopandas as gpd

url = (
    "https://spatial.infrastructure.gov.au/server/rest/services/"
    "Hosted/Com_BushTel_Profile_CMC_2024/FeatureServer/0/query"
    "?where=1%3D1"
    "&outFields=%2A"
    "&returnGeometry=true"
    "&f=geojson"
)

gdf = gpd.read_file(url)

print(gdf.shape)
display(gdf.head())

# Lưu GeoJSON
gdf.to_file("bushtel_communities_2024.geojson", driver="GeoJSON")

# Lưu CSV
gdf.drop(columns="geometry").to_csv(
    "bushtel_communities_2024.csv",
    index=False
)
```
